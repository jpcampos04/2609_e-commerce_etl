
## Role: Data Engineers (DEs)
### Complete access: can create/modify objects in the 3 layers

```python
%sql
-- CATALOG
GRANT ALL PRIVILEGES ON CATALOG catalog_au TO `DEs`;
 
-- GRANT SCHEMA
GRANT CREATE, USE SCHEMA ON SCHEMA catalog_au.bronze TO `DEs`;
GRANT CREATE, USE SCHEMA ON SCHEMA catalog_au.silver TO `DEs`;
GRANT CREATE, USE SCHEMA ON SCHEMA catalog_au.gold   TO `DEs`;
 
-- GRANT TABLES — bronze
GRANT SELECT ON TABLE catalog_au.bronze.customers   TO `DEs`;
GRANT SELECT ON TABLE catalog_au.bronze.geolocation TO `DEs`;
GRANT SELECT ON TABLE catalog_au.bronze.items       TO `DEs`;
GRANT SELECT ON TABLE catalog_au.bronze.orders      TO `DEs`;
GRANT SELECT ON TABLE catalog_au.bronze.payments    TO `DEs`;
GRANT SELECT ON TABLE catalog_au.bronze.products    TO `DEs`;
GRANT SELECT ON TABLE catalog_au.bronze.sellers     TO `DEs`;
GRANT SELECT ON TABLE catalog_au.bronze.translation TO `DEs`;
 
-- GRANT TABLES — silver
GRANT SELECT ON TABLE catalog_au.silver.orders_transformed TO `DEs`;
 
-- GRANT TABLES — gold
GRANT SELECT ON TABLE catalog_au.gold.gold_sales_monthly        TO `DEs`;
GRANT SELECT ON TABLE catalog_au.gold.gold_delivery_performance TO `DEs`;
GRANT SELECT ON TABLE catalog_au.gold.gold_geo_concentration    TO `DEs`;
 
-- GRANT EXTERNAL LOCATION
GRANT READ FILES, WRITE FILES ON EXTERNAL LOCATION `raw_ext`    TO `DEs`;
GRANT READ FILES, WRITE FILES ON EXTERNAL LOCATION `bronze_ext` TO `DEs`;
GRANT READ FILES, WRITE FILES ON EXTERNAL LOCATION `silver_ext` TO `DEs`;
GRANT READ FILES, WRITE FILES ON EXTERNAL LOCATION `gold_ext`   TO `DEs`;
```
